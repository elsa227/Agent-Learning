多轮对话

```python
from langchain_core.messages import HumanMessage
from langchain_core.messages import AIMessage
from langchain_core.messages import SystemMessage

messages = [
    SystemMessage(content="你是一个专业的Python编程助手，擅长解释技术概念"),
]

print("多轮对话示例（输入‘quit’退出\n）")

while True:
    user_input = input("你: ")
    if user_input.lower() == "quit":
        break

    #添加用户信息
    messages.append(HumanMessage(content=user_input))

    #调用模型
    response = model.invoke(messages)

    #添加AI回复到历史
    messages.append(AIMessage(content=response.content))

    print(f"AI: {response.content}")
```

流式输出

```python
print("流式输出示例\n")
question = "请写一个Python冒泡排序的实现"
print(f"用户:{question}\n")
print("AI:",end="",flush=True)

for trunk in model.stream(question):
    print(trunk.content,end="",flush=True)

print("\n")
```

```python
from langchain.agents import create_agent
from langchain_core.tools import tool

# 1. 定义一个普通函数，加上 @tool 装饰器
@tool
def get_weather(city: str) -> str:
    """获取指定城市的天气情况"""
    # 这里假装查了天气，实际开发中会用 requests 去调真实的天气 API
    return f"{city}的天气是晴天"

# 2. 创建 Agent（把大模型和工具组装起来）
agent = create_agent(
    model=model,             # 传入你之前初始化好的大模型
    tools=[get_weather],     # 告诉 Agent 它有这个工具可以用
    system_prompt="你是一个有用的助手" # 给 AI 的底层人设
)

# 3. 发起提问，看 AI 怎么处理
result = agent.invoke(
    {"messages": [{"role": "user", "content": "北京天气怎么样？"}]}
)

# 4. 打印结果（这里建议只打印 AI 的最终回答，不然会出来一堆字典）
print(result["messages"][-1].content)
```

实战：智能翻译助手

```python
import os
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
from langchain_core.messages import SystemMessage, HumanMessage

# 加载环境变量
load_dotenv()

class SmartTranslator:
    # ==========================================
    # 1. 初始化方法
    # ==========================================
    def __init__(self):
        # 从环境变量中读取硅基流动 Key
        api_key = os.getenv("SILICONFLOW_API_KEY") 
        
        # 初始化模型
        self.model = init_chat_model(
            "Qwen/Qwen2.5-7B-Instruct",  # 稳妥起见，这里我直接帮你换成了肯定免费的 2.5-7B
            model_provider="openai",
            base_url="https://api.siliconflow.cn/v1",
            api_key=api_key,
            temperature=0.3  # 翻译任务使用较低温度，保证严谨
        )

    # ==========================================
    # 2. 核心翻译方法（注意这里的缩进，要和 __init__ 对齐）
    # 定义了翻译的动作。必须给它 text（要翻译的词），顺便给了默认值：
    # 如果用户不说，就默认翻译成“中文”、“正式”风格。
    # ==========================================
    def translate(self, text: str, target_lang: str = "中文", style: str = "正式"):
        """
        翻译文本

        Args:
            text: 要翻译的文本
            target_lang: 目标语言（中文/英文/日文等）
            style: 翻译风格（正式/口语/文学）
        """
        system_prompt = f"""你是一个专业的翻译助手。

            任务：
            1. 自动检测输入文本的语言
            2. 翻译成{target_lang}
            3. 使用{style}风格
            4. 如果有专业术语，在翻译后用括号标注原文

            输出格式：
            【原语言】：xxx
            【翻译】：xxx
            【术语解释】：（如果有）
        """

        # 构建消息列表
        messages = [
            SystemMessage(content=system_prompt),
            HumanMessage(content=text)
        ]

        # 调用模型并返回文本结果
        response = self.model.invoke(messages)
        return response.content

# ==========================================
# 3. 主程序运行入口
# ==========================================
def main():
    # 实例化翻译官
    translator = SmartTranslator()

    print("🌍 智能翻译助手 (LangChain 1.0)")
    print("=" * 50)

    # 示例1: 英译中 (技术文本)
    text1 = "LangChain is a framework for developing applications powered by large language models."
    print(f"\n📝 原文: {text1}")
    # 注意这里调用了 translator 实例的 translate 方法
    print(f"\n🔄 翻译结果:\n{translator.translate(text1, '中文', '正式')}")
    print("\n" + "=" * 50)

    # 示例2: 中译英 (口语风格)
    text2 = "这个框架真的超级好用，强烈推荐！"
    print(f"\n📝 原文: {text2}")
    print(f"\n🔄 翻译结果:\n{translator.translate(text2, '英文', '口语')}")
    print("\n" + "=" * 50)

    # 交互模式
    print("\n💬 进入交互模式 (输入 'quit' 退出) \n")
    while True:
        text = input("请输入要翻译的文本：")
        if text.lower() == 'quit':
            break

        target = input("目标语言 (默认中文)：") or "中文"
        style = input("翻译风格 (正式/口语/文学, 默认正式)：") or "正式"

        print(f"\n🔄 翻译中...\n")
        result = translator.translate(text, target, style)
        print(result)
        print("\n" + "=" * 50 + "\n")

# Python 标准入口
if __name__ == "__main__":
    main()
```
